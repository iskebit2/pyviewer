```python
# windplane.py

import math
import numpy as np
from dataclasses import dataclass
from enum import Enum
from typing import List, Dict, Tuple, Optional, Union, Any
from data.wind_data import ARAZI_KATEGORILERI, CPI_POSITIVE, CPI_NEGATIVE, RHO,K_I,C0, wall_cpe_table,MONOPITCH_CPE_DATA,DUOPITCH_CPE_DATA,HIPPED_CPE_DATA

class SurfaceType(Enum):
    ROOF = "ROOF"
    WALL = "WALL"

class WindRelation(Enum):
    nodata = "nodata"
    WINDWARD = "WINDWARD"
    LEEWARD = "LEEWARD"
    PARALLEL = "PARALLEL"

EPS = 1e-8


@dataclass
class Zone:
    name: str
    polygon: np.ndarray
    cpe_neg: Optional[float] = None
    cpe_pos: Optional[float] = None


def close_polygon(polygon: np.ndarray, tol: float = 1e-8) -> np.ndarray:
    """Poligonu kapatır"""
    polygon = np.asarray(polygon, dtype=float)
    if len(polygon) == 0:
        return polygon
    if not np.allclose(polygon[0], polygon[-1], atol=tol):
        polygon = np.vstack([polygon, polygon[0]])
    return polygon


def same_edge(edge1: Dict, edge2: Dict, tol: float = 1e-6) -> bool:
    """İki kenarın aynı olup olmadığını kontrol eder"""
    p1 = np.asarray(edge1["p1"], dtype=float)
    p2 = np.asarray(edge1["p2"], dtype=float)

    q1 = np.asarray(edge2["p1"], dtype=float)
    q2 = np.asarray(edge2["p2"], dtype=float)

    return (
        np.allclose(p1, q1, atol=tol) and np.allclose(p2, q2, atol=tol)
    ) or (
        np.allclose(p1, q2, atol=tol) and np.allclose(p2, q1, atol=tol)
    )

def same_axis_xy(
    edge1: dict,
    edge2: dict,
    angle_tol: float = 2.0,
    distance_tol: float = 1e-6
) -> bool:

    d1 = edge1["direction_xy"]
    d2 = edge2["direction_xy"]

    if d1 is None or d2 is None:
        return False

    # 1. Doğrultular paralel mi?
    dot = np.clip(abs(np.dot(d1, d2)), -1.0, 1.0)
    angle = np.degrees(np.arccos(dot))

    if angle > angle_tol:
        return False

    # 2. Aynı aks üzerindeler mi?
    p1 = edge1["p1"][:2]
    q1 = edge2["p1"][:2]

    # d1'e dik birim vektör
    normal = np.array([-d1[1], d1[0]])

    distance = abs(np.dot(q1 - p1, normal))

    return distance <= distance_tol

def polygon_edges(polygon: np.ndarray) -> List[Dict]:
    """Poligonun kenarlarını döndürür"""
    polygon = np.asarray(polygon, dtype=float)

    if np.allclose(polygon[0], polygon[-1]):
        polygon = polygon[:-1]

    n = len(polygon)
    edges = []

    for i in range(n):
        edges.append({
            "index": i,
            "p1": polygon[i],
            "p2": polygon[(i + 1) % n],
        })

    return edges


def polygon_centroid(polygon: np.ndarray) -> np.ndarray:
    """Poligonun merkezini hesaplar"""
    pts = np.asarray(polygon, dtype=float)

    if np.allclose(pts[0], pts[-1]):
        pts = pts[:-1]

    return np.mean(pts, axis=0)

def dict_tree(data, indent=""):
    lines = []

    items = list(data.items())

    for i, (key, value) in enumerate(items):
        last = i == len(items) - 1

        branch = "└── " if last else "├── "
        lines.append(f"{indent}{branch}{key}")

        if isinstance(value, dict):
            new_indent = indent + ("    " if last else "│   ")
            lines.append(dict_tree(value, new_indent))

        else:
            # Yukarıdaki key satırını value ile birleştir
            lines[-1] = f"{indent}{branch}{key} : {value}"

    return "\n".join(lines)
```

```python
# edge_analyzer.py

import numpy as np
from typing import Dict, Any, Optional, List, Tuple


class EdgeAnalyzer:
    """
    Bir WindPlane'in kenarlarını, verilen rüzgar yönü için XY izdüşümünde
    analiz eder.

    Temel ilkeler
    -------------
    1. Tüm geometri 2D XY düzlemindedir. Rüzgar vektörü [x, y, 0] kabul edilir.
    2. Kenar "exposed" tespiti DÖNME KURALI ile yapılır:
         - Poligon CCW ise, kenar boyunca ilerlerken sol taraf İÇ, sağ taraf DIŞ.
         - Poligon CW ise, sağ taraf İÇ, sol taraf DIŞ.
         - Rüzgar vektörü (gittiği yön) kenarın DIŞ tarafına düşüyorsa,
           kenar rüzgarı karşılıyor demektir → exposed.
         - 2D cross product:  cross = e × w
             cross < 0  →  w, e'nin sağında
             cross > 0  →  w, e'nin solunda
         - CCW:  exposed  ⟺  cross < 0
         - CW :  exposed  ⟺  cross > 0
    3. Leading kenar: exposed kenarlar arasında rüzgar ekseninde EN ÖNDE
       başlayan (en küçük pos_front) kenar(lar).
    4. shared / same_axis: Diğer çatı yüzeyleriyle geometrik karşılaştırma.

    Girdi Sözleşmesi
    ----------------
    edges[i] = {
        "p1": np.array([x, y, z]),
        "p2": np.array([x, y, z]),
        "vector": np.array([dx, dy, dz]),
        "length": float,
        "unit_vector": np.array([...]),
        "vector_xy": np.array([dx, dy]),
        "length_xy": float,
        "direction_xy": np.array([dx, dy]) veya None,   # birim, XY
    }

    Çıktı Sözleşmesi
    ----------------
    Aynı sözlük + şu alanlar:
        pos_front : float   # rüzgar ekseninde ön uç pozisyonu (küçük)
        pos_back  : float   # rüzgar ekseninde arka uç pozisyonu (büyük)
        angle     : float   # kenar ile rüzgarın yaptığı 2D açı (derece)
        exposed   : bool    # dönme kuralına göre rüzgara maruz
        leading   : bool    # exposed + global ön hatta
        vertical  : bool    # XY'de sıfır uzunluklu (tam düşey) kenar
        shared    : list    # diğer çatı yüzeyleriyle tam çakışan kenarlar
        same_axis : list    # aynı doğru üzerindeki diğer kenarlar
    """

    def __init__(
        self,
        edges: Dict[int, Dict[str, Any]],
        surface_normal: np.ndarray,
        is_ccw: bool,
        tol: float = 1e-6,
    ):
        self.edges = edges
        self.surface_normal = np.asarray(surface_normal, dtype=float)
        self.is_ccw = bool(is_ccw)
        self.tol = float(tol)

        # analyze() çağrısında doldurulur
        self._wind_from_2d: Optional[np.ndarray] = None   # rüzgarın geldiği yön
        self._wind_to_2d: Optional[np.ndarray] = None     # rüzgarın gittiği yön
        self._analyzed: Dict[int, Dict[str, Any]] = {}

    # ------------------------------------------------------------------
    # PUBLIC API
    # ------------------------------------------------------------------

    def analyze(
        self,
        wind_vector: np.ndarray,
        building: Any = None,
        current_surface_name: str = "",
    ) -> Dict[int, Dict[str, Any]]:
        """
        Ana giriş noktası.

        Args:
            wind_vector: 3B vektör [wx, wy, 0]. Z bileşeni yok sayılır.
            building: BuildingWindEngine (surfaces_items içerir). None ise
                      yalnızca tek yüzey analizi yapılır.
            current_surface_name: Bu yüzeyin adı (kendini hariç tutmak için).

        Returns:
            {edge_id: zenginleştirilmiş kenar sözlüğü}
        """
        w_to = self._normalize_wind(wind_vector)
        if w_to is None:
            return {}

        self._wind_to_2d = w_to              # rüzgarın gittiği yön
        self._wind_from_2d = -w_to           # rüzgarın geldiği yön

        # 1) Her kenar için 2D metrikler + exposed (dönme kuralı)
        self._analyzed = self._compute_per_edge_metrics()
        if not self._analyzed:
            return {}

        # 2) Leading tespiti (yüzey içi)
        self._mark_leading()

        # 3) Diğer çatılarla karşılaştırma
        if building is not None:
            self._find_shared_and_same_axis(building, current_surface_name)

        return self._analyzed

    # ------------------------------------------------------------------
    # YARDIMCI SORGULAR
    # ------------------------------------------------------------------

    def leading_edges(self) -> Dict[int, Dict[str, Any]]:
        return {i: e for i, e in self._analyzed.items() if e["leading"]}

    def exposed_edges(self) -> Dict[int, Dict[str, Any]]:
        return {i: e for i, e in self._analyzed.items() if e["exposed"]}

    def shared_edges(self) -> Dict[int, Dict[str, Any]]:
        return {i: e for i, e in self._analyzed.items() if e["shared"]}

    def has_leading(self) -> bool:
        return any(e["leading"] for e in self._analyzed.values())

    # ------------------------------------------------------------------
    # 1) RÜZGAR NORMALİZASYONU
    # ------------------------------------------------------------------

    @staticmethod
    def _normalize_wind(wind_vector: np.ndarray) -> Optional[np.ndarray]:
        """Rüzgar vektörünün XY bileşenini birim vektör yapar."""
        w = np.asarray(wind_vector, dtype=float)
        w2 = w[:2]
        n = np.linalg.norm(w2)
        if n < 1e-12:
            return None
        return w2 / n

    # ------------------------------------------------------------------
    # 2) KENAR BAŞINA 2D METRİKLER + EXPOSED
    # ------------------------------------------------------------------

    def _compute_per_edge_metrics(self) -> Dict[int, Dict[str, Any]]:
        result: Dict[int, Dict[str, Any]] = {}

        for edge_id, base in self.edges.items():
            d2 = base.get("direction_xy", None)

            # Düşey kenar (XY'de sıfır uzunluk)
            if d2 is None:
                result[edge_id] = self._vertical_edge_entry(base)
                continue

            d2 = np.asarray(d2, dtype=float)
            p1_2d = base["p1"][:2]
            p2_2d = base["p2"][:2]

            # --- Rüzgar ekseninde ön/arka pozisyon ---
            s1 = float(np.dot(p1_2d, self._wind_from_2d))
            s2 = float(np.dot(p2_2d, self._wind_from_2d))
            pos_front, pos_back = (s1, s2) if s1 <= s2 else (s2, s1)

            # --- Kenar ile rüzgar arasındaki açı (2D) ---
            cos_a = float(np.clip(abs(np.dot(d2, self._wind_from_2d)), -1.0, 1.0))
            angle = float(np.degrees(np.arccos(cos_a)))

            # --- EXPOSED: DÖNME KURALI ---
            exposed = self._is_exposed(base)

            result[edge_id] = {
                **base,
                "pos_front": pos_front,
                "pos_back": pos_back,
                "angle": angle,
                "exposed": exposed,
                "leading": False,      # sonra doldurulur
                "vertical": False,
                "shared": [],
                "same_axis": [],
            }

        return result

    def _is_exposed(self, edge: Dict[str, Any]) -> bool:
        """
        Dönme kuralı ile exposed tespiti.

        Kenar vektörü e = p2 - p1 (XY).
        Rüzgar vektörü w = wind_to_2d (gittiği yön).
        cross = e.x * w.y - e.y * w.x

        CCW poligonda:
            cross < 0  →  rüzgar kenarın sağında (dış taraf)  →  exposed
            cross > 0  →  rüzgar kenarın solunda (iç taraf)   →  değil

        CW poligonda:
            cross > 0  →  rüzgar kenarın sağında (dış taraf)  →  exposed
            cross < 0  →  rüzgar kenarın solunda (iç taraf)   →  değil
        """
        e = np.asarray(edge["p2"][:2] - edge["p1"][:2], dtype=float)
        w = self._wind_to_2d

        cross = e[0] * w[1] - e[1] * w[0]

        if self.is_ccw:
            return cross < -self.tol
        else:
            return cross > self.tol

    def _vertical_edge_entry(self, base: Dict[str, Any]) -> Dict[str, Any]:
        """
        XY'de uzunluğu sıfır olan (tam düşey) kenar.
        Rüzgar karşılama testine girmez; exposed = False.
        """
        p_2d = base["p1"][:2]
        pos = float(np.dot(p_2d, self._wind_from_2d))

        return {
            **base,
            "pos_front": pos,
            "pos_back": pos,
            "angle": 90.0,       # XY'de düşey → rüzgarla tanımsız, 90° kabul
            "exposed": False,
            "leading": False,
            "vertical": True,
            "shared": [],
            "same_axis": [],
        }

    # ------------------------------------------------------------------
    # 3) LEADING TESPİTİ
    # ------------------------------------------------------------------

    def _mark_leading(self) -> None:
        """
        Exposed kenarlar arasında EN ÖN pozisyona (min pos_front) sahip
        olanlar leading olur.

        İki kenar aynı ön köşeyi paylaşıyorsa, ikisinin de pos_front değeri
        aynıdır ve ikisi de leading işaretlenir.
        """
        exposed_edges = [e for e in self._analyzed.values() if e["exposed"]]
        if not exposed_edges:
            return

        global_front = min(e["pos_front"] for e in exposed_edges)

        for e in exposed_edges:
            if abs(e["pos_front"] - global_front) <= self.tol:
                # Nokta teması değil, gerçek kenar
                if e["length_xy"] > 1e-3:
                    e["leading"] = True

    # ------------------------------------------------------------------
    # 4) DİĞER YÜZEYLERLE KARŞILAŞTIRMA
    # ------------------------------------------------------------------

    def _find_shared_and_same_axis(
        self,
        building: Any,
        current_surface_name: str,
    ) -> None:
        """
        Bu yüzeyin kenarlarını, building içindeki diğer çatı yüzeylerinin
        kenarlarıyla karşılaştırır.
        """
        surfaces_items = getattr(building, "surfaces_items", {}) or {}

        other_roofs: List[Tuple[str, Any]] = []
        for name, surf in surfaces_items.items():
            if name == current_surface_name:
                continue
            stype = getattr(surf.surface_type, "value", surf.surface_type)
            if stype != "ROOF":
                continue
            other_roofs.append((name, surf))

        if not other_roofs:
            return

        for edge in self._analyzed.values():
            if edge.get("direction_xy", None) is None:
                continue

            for other_name, other_surf in other_roofs:
                for other_id, other_edge in other_surf.edges.items():
                    if not self._same_axis_xy(edge, other_edge):
                        continue

                    is_same = self._same_edge(edge, other_edge)
                    entry = {
                        "surface": other_name,
                        "surface_edge": other_id,
                        "surface_angle": getattr(other_surf, "angle", None),
                        "direction_xy": other_edge["direction_xy"],
                        "same_edge": is_same,
                    }
                    edge["same_axis"].append(entry)
                    if is_same:
                        edge["shared"].append(entry)

    def _same_axis_xy(self, e1: Dict[str, Any], e2: Dict[str, Any]) -> bool:
        """İki kenar aynı XY aksı üzerinde mi? (paralel + aynı doğru)"""
        d1 = e1.get("direction_xy", None)
        d2 = e2.get("direction_xy", None)
        if d1 is None or d2 is None:
            return False

        dot = float(np.clip(abs(np.dot(d1, d2)), -1.0, 1.0))
        if np.degrees(np.arccos(dot)) > 2.0:
            return False

        p1 = e1["p1"][:2]
        q1 = e2["p1"][:2]
        normal = np.array([-d1[1], d1[0]])
        return abs(float(np.dot(q1 - p1, normal))) <= self.tol

    def _same_edge(self, e1: Dict[str, Any], e2: Dict[str, Any]) -> bool:
        """İki kenar tamamen çakışık mı?"""
        p1, p2 = e1["p1"], e1["p2"]
        q1, q2 = e2["p1"], e2["p2"]
        return (
            np.allclose(p1, q1, atol=self.tol)
            and np.allclose(p2, q2, atol=self.tol)
        ) or (
            np.allclose(p1, q2, atol=self.tol)
            and np.allclose(p2, q1, atol=self.tol)
        )
```

```python


class WindPlane:
    def __init__(self, polygon: Union[List, np.ndarray], name: str = "Test"):
        """3B Düzlemsel Poligon Nesnesi"""
        pts = np.asarray(polygon, dtype=float)

        if len(pts) > 1 and np.allclose(pts[0], pts[-1], atol=1e-8):
            pts = pts[:-1]

        if len(pts) < 3:
            raise ValueError("Poligon en az 3 noktadan oluşmalıdır.")

        self.polygon_name = name
        self.coords = pts
        self.pts_3d: np.ndarray = close_polygon(pts) # Harici helper
        self.n_pts: int = len(pts)

        self.normal, self.normal_unit = self._compute_normal()
        self.centroid = polygon_centroid(self.pts_3d) # Harici helper
        self.pitch: float = self._compute_pitch()
        self.surface_type: SurfaceType = SurfaceType.ROOF if self.pitch <= 75.0 else SurfaceType.WALL
        self.proj_info = self._get_best_projection_plane()
        self.angle = self._compute_angle()
        self.exposed_edge_list = []
        self.edges = self._build_edges()
        self.global_leading = False

        self.properties = {
            "name": self.polygon_name,
            "polygon": self.pts_3d,
            "surface_type": self.surface_type,
            "angle": self.angle,
            "pitch": self.pitch,
            "coords": pts,
            "zmin": float(np.min(pts[:, 2])),
            "zmax": float(np.max(pts[:, 2])),
            "wind_vector": None,
            "wind_relation": WindRelation.nodata,
            "global_leading": None
        }

    # --- GEOMETRİK HESAPLAMALAR ---

    def _compute_normal(self) -> Tuple[np.ndarray, np.ndarray]:
        normal = np.zeros(3)
        for i in range(self.n_pts):
            p_curr, p_next = self.pts_3d[i], self.pts_3d[i + 1]
            normal[0] += (p_curr[1] - p_next[1]) * (p_curr[2] + p_next[2])
            normal[1] += (p_curr[2] - p_next[2]) * (p_curr[0] + p_next[0])
            normal[2] += (p_curr[0] - p_next[0]) * (p_curr[1] + p_next[1])

        norm_len = np.linalg.norm(normal)
        if norm_len < 1e-12:
            raise ValueError(f"Geçersiz/Çökmüş poligon düzlemi: {self.pts_3d}")
        return normal, normal / norm_len

    def _compute_pitch(self) -> float:
        nz_ratio = abs(self.normal_unit[2])
        nz_ratio = np.clip(nz_ratio, -1.0, 1.0)
        return float(np.degrees(np.arccos(nz_ratio)))

    def _compute_angle(self) -> float:
        if self.pitch < 1e-10:
            return 0.0
        nx, ny, nz = self.normal_unit
        if self.pitch >= 90.0 - 1e-10:
            return float(np.degrees(np.arctan2(ny, nx)))
        horiz_len = np.hypot(nx, ny)
        if horiz_len < 1e-12:
            return 0.0
        sign = np.sign(ny) if abs(ny) > abs(nx) else np.sign(nx)      
        return float(sign * self.pitch)

    def _get_best_projection_plane(self):
        abs_n = np.abs(self.normal_unit)
        max_idx = np.argmax(abs_n)
        if max_idx == 2:
            return 'XY', (0, 1), 2
        elif max_idx == 1:
            return 'XZ', (0, 2), 1
        else:
            return 'YZ', (1, 2), 0

    def polygon_direction_xy(self) -> str:
        """Poligonun XY düzlemindeki yönünü belirler (CW / CCW)."""
        polygon = self.coords
        n = len(polygon)
        area2 = 0.0
        for i in range(n):
            x1, y1 = polygon[i][:2]
            x2, y2 = polygon[(i + 1) % n][:2]
            area2 += (x1 * y2 - x2 * y1)

        if area2 > EPS:
            return "CCW"
        elif area2 < -EPS:
            return "CW"
        return "DEGENERATE"

    def _build_edges(self) -> dict:
        edges = {}
        for i in range(self.n_pts):
            p1 = self.pts_3d[i]
            p2 = self.pts_3d[i + 1]

            vector = p2 - p1
            length = np.linalg.norm(vector)
            if length < 1e-12:
                continue

            unit_vector = vector / length
            vector_xy = vector[:2]
            length_xy = np.linalg.norm(vector_xy)
            direction_xy = vector_xy / length_xy if length_xy >= 1e-12 else None

            edges[i] = {
                "p1": p1.copy(),
                "p2": p2.copy(),
                "vector": vector.copy(),
                "length": float(length),
                "unit_vector": unit_vector.copy(),
                "vector_xy": vector_xy.copy(),
                "length_xy": float(length_xy),
                "direction_xy": direction_xy,
                "shared": [],
                "same_axis": [],
                "exposed": False,
                "leading": False
            }
        return edges

    # --- RÜZGAR VE KENAR ANALİZİ ---

    def analyze_wind_relation(self, w_vector: np.ndarray) -> WindRelation:
        w_xy = np.asarray(w_vector[:2], dtype=float)
        w_xy /= np.linalg.norm(w_xy)
        n_xy = self.normal_unit[:2]
        dot = np.dot(n_xy, w_xy)
    
        if dot < -0.2: return WindRelation.WINDWARD
        elif dot > 0.2: return WindRelation.LEEWARD
        else: return WindRelation.PARALLEL
    
    def _analyze_edges(self, w: np.ndarray, building=None) -> Dict[int, Dict[str, Any]]:
        """Kenar analizini EdgeAnalyzer nesnesine devreder (Delegation)."""
        is_ccw = (self.polygon_direction_xy() == "CCW")
    
        analyzer = EdgeAnalyzer(
            edges=self.edges,
            surface_normal=self.normal_unit,
            is_ccw=is_ccw,
            tol=1e-6,
        )
    
        return analyzer.analyze(
            wind_vector=w,
            building=building,
            current_surface_name=self.polygon_name,
        )
        

    def _check_geometry_shape(self) -> bool:
        """Yüzeyin üçgen veya yamuk (trapezoidal) olup olmadığını tespit eder."""
        n_pts = len(self.coords)
        if n_pts == 3:
            return True
        elif n_pts == 4:
            # Karşılıklı kenar vektörlerinin paralelliğini kontrol et
            v0 = self.coords[1] - self.coords[0]
            v1 = self.coords[2] - self.coords[1]
            v2 = self.coords[3] - self.coords[2]
            v3 = self.coords[0] - self.coords[3]
            
            # Cross product 0'a yakınsa kenarlar paraleldir
            cross1 = np.linalg.norm(np.cross(v0, v2))
            cross2 = np.linalg.norm(np.cross(v1, v3))
            if cross1 < 1e-4 or cross2 < 1e-4:
                return True
        return False

    def _detect_ridge_neighbor(self) -> bool:
        """Yüzeyin en yüksek kotundaki kenarlarında ortak mahya komşusu var mı taranır."""
        z_max_surface = self.properties["zmax"]
        for edge_id, edge in self.edges.items():
            # Kenar mahya kotuna yakın mı?
            edge_z_max = max(edge["p1"][2], edge["p2"][2])
            if abs(edge_z_max - z_max_surface) < 1e-3:
                # Ortak komşusu olan bir çatı yüzeyi var mı?
                if len(edge["shared"]) > 0:
                    return True
        return False

    def _check_gable_side(self, w: np.ndarray) -> bool:
        """Beşik çatı yan yüzeyi durumunu kontrol eder."""
        if self.surface_type != SurfaceType.ROOF:
            return False
        # Üçgen/Yamuk değilse ve mahya komşuluğu varsa beşik çatı yan yüzeyidir
        is_shape_tri_trap = self._check_geometry_shape()
        has_ridge = self._detect_ridge_neighbor()
        return (not is_shape_tri_trap) and has_ridge

    def _check_asymmetric_gable(self) -> dict:
        """X veya Y düzlemine paralel kalkan duvar kesitlerini tespit eder."""
        has_gable_cut = False
        for e in self.edges.values():
            v = e["vector"]
            # Kenar X veya Y ekseninde sabit kalıp Z kotunda yükseliyorsa kalkan kenarıdır
            is_const_x = abs(v[0]) < 1e-3
            is_const_y = abs(v[1]) < 1e-3
            has_z_slope = abs(v[2]) > 1e-3
    
            if (is_const_x or is_const_y) and has_z_slope:
                has_gable_cut = True
                break
                
        return {
            "has_gable_cut": has_gable_cut,
            "is_pure_hip": self._check_geometry_shape() and not has_gable_cut
        }
        
    def analysis_(self, w, building) -> None:
        rel_ = self.analyze_wind_relation(w)
    
        # EdgeAnalyzer ile kenar analizi
        new_edges = self._analyze_edges(w, building)
        self.edges = new_edges                      # ← referansı tazele
    
        self.wind_relation = rel_                   # ← attribute ata
        self.properties["wind_relation"] = rel_
        self.properties["wind_vector"] = w
        self.properties["any_shared"] = any(
            e.get("shared") for e in self.edges.values()
        )
    
        if self.surface_type.value == "WALL":
            self.properties["regions"] = self.calculate_wall_cpe_from_polygon(
                w, building.geometry
            )
```

```python
class BuildingWindEngine:
    

    def __init__(
        self, 
        points: Dict[str, Tuple[float, float, float]], 
        polygons: Dict[str, List[str]], 
        v_b0: float = 28.0, 
        terrain: str = "Kategori III",
        w_dir: List[float] = [1.0, 0.0, 0.0],
        scale_factor: float = 1000.0,
        rho: float = 1.25,
        ct: float = 1.0,
        kI: float = 1.0
    ):
        self.raw_points = points
        self.polygons = polygons
        self.v_b0 = float(v_b0)
        self.terrain = terrain
        self.w_dir = np.array(w_dir, dtype=float)
        self.scale_factor = float(scale_factor)
        self.rho = rho
        self.ct = ct
        self.kI = kI
        self.e= None
        self.all_roof_polygons = {}

        if self.terrain not in ARAZI_KATEGORILERI:
            raise ValueError(f"Geçersiz arazi kategorisi: {self.terrain}")
        
        self.z0 = ARAZI_KATEGORILERI[self.terrain]["z0"]
        self.zmin = ARAZI_KATEGORILERI[self.terrain]["zmin"]

        # 1. Noktaları ölçekle
        self.scaled_points = self._scale_points()
        
        # 2. Yüzeyleri analiz et (Duvar/Çatı ayrımı ve kotlar)
        self.surfaces = self._analyze_surfaces()
        
        # 3. Yüzeylerden gelen verilerle bina geometrisini bağla (Artık çakışma yok)
        self.geometry = self._calculate_building_geometry()
        
        # 4. Rüzgar parametrelerini (q_b, q_p) hesapla
        self._calculate_wind_parameters()

    def _scale_points(self) -> Dict[str, np.ndarray]:
        return {
            name: np.array(pt, dtype=float) / self.scale_factor 
            for name, pt in self.raw_points.items()
        }

    def _analyze_surfaces(self) -> Dict[str, Dict[str, Any]]:
        """Poligon bağımsız yüzey tipini ve kotlarını çıkarır."""
        surfaces = {}
        self.surfaces_items= {}
        self.all_roof_polygons = {}
        for poly_name, pt_names in self.polygons.items():
            pts = np.array([self.scaled_points[pt] for pt in pt_names])
            surface_= WindPlane(pts, name=poly_name)
            self.surfaces_items[poly_name] = surface_
            if surface_.surface_type.value=="ROOF":
                self.all_roof_polygons[poly_name]= surface_.properties['coords']


            surfaces[poly_name] = {
                "type": surface_.properties['surface_type'],
                "pitch": surface_.properties['pitch'],
                "relation": surface_.analyze_wind_relation(self.w_dir),
                "zmin": surface_.properties['zmin'],
                "zmax": surface_.properties['zmax'],
                "coords": surface_.properties['coords']
            }
        return surfaces

    def _calculate_building_geometry(self) -> Dict[str, float]:
        """Hazır self.surfaces üzerinden saçak kotunu çekerek geometriyi tamamlar."""
        pts_matrix = np.array(list(self.scaled_points.values()))

        # Rüzgar eksenleri
        w_xy = self.w_dir[:2] / np.linalg.norm(self.w_dir[:2])
        v_vec = np.array([-w_xy[1], w_xy[0]])

        u_vals = [np.dot(pt[:2], w_xy) for pt in pts_matrix]
        v_vals = [np.dot(pt[:2], v_vec) for pt in pts_matrix]

        b = max(v_vals) - min(v_vals)
        d = max(u_vals) - min(u_vals)
        
        z_ground = float(min(pt[2] for pt in pts_matrix))
        z_ridge = float(max(pt[2] for pt in pts_matrix))

        # Saçak kotu tespiti (Duvarların maksimum z kotu)
        wall_z_maxs = [s["zmin"] for s in self.surfaces.values() if s["type"] == SurfaceType.ROOF]
        z_eaves = max(wall_z_maxs) if wall_z_maxs else z_ridge

        h_total = z_ridge - z_ground    # Mahya bazlı bina toplam yüksekliği
        h_eaves = z_eaves - z_ground    # Saçak yüksekliği
        e = min(b, 2.0 * h_total)
        self.e= e
        return {
            "b": b,
            "d": d,
            "h": h_total,
            "h_eaves": h_eaves,
            "z0_ground": z_ground,
            "z1_eaves": z_eaves,
            "z2_ridge": z_ridge,
            "e": e
        }

    def _calculate_wind_parameters(self):
        self.q_b = 0.5 * self.rho * (self.v_b0 ** 2) / 1000.0
        self.z_ref = max(self.geometry["z2_ridge"], self.zmin)

        self.kr = 0.19 * ((self.z0 / 0.05) ** 0.07)
        self.cr = self.kr * math.log(self.z_ref / self.z0)
        self.Iv = self.kI / math.log(self.z_ref / self.z0)

        self.ce = (self.cr * self.ct) ** 2 * (1.0 + 7.0 * self.Iv)
        self.q_p = self.ce * self.q_b

    def get_summary(self) -> Dict[str, Any]:
        """Tüm özet sonuçları temiz bir dict olarak döndürür."""
        return {
            "Geometri": self.geometry,
            "Rüzgar Parametreleri": {
                "q_b (kN/m²)": round(self.q_b, 3),
                "q_p (kN/m²)": round(self.q_p, 3),
                "c_e": round(self.ce, 3),
                "I_v": round(self.Iv, 3),
                "c_r": round(self.cr, 3),
                "k_r": round(self.kr, 3),
                "z_ref (m)": round(self.z_ref, 3),
            },
            "Arazi": {
                "Kategori": self.terrain,
                "z0 (m)": self.z0,
                "zmin (m)": self.zmin
            }
        }

    def get_polygon_coords(self, polygon_points, points, scale_=1000.0):
        """Poligon noktalarını koordinatlara dönüştürür ve ölçekler."""
        coords = []
        for pt_name in polygon_points:
            if pt_name in points:
                pt = points[pt_name]
                coords.append([pt[0] / scale_, pt[1] / scale_, pt[2] / scale_])
            else:
                raise ValueError(f"Nokta '{pt_name}' bulunamadı!")
        return np.array(coords)

    def analysis_all_roof_wind(self, w):
        w = np.array(w, dtype=float)
        w_len = np.linalg.norm(w)
        if w_len < 1e-12:
            return
        w_norm = w / w_len

        roof_surfaces = {}
        for surface_name, surface in self.surfaces_items.items():
            if getattr(surface.surface_type, "value", surface.surface_type) == "ROOF":
                surface.analysis_(w, self)
                surface.global_leading = False
                surface.properties["global_leading"] = False
                roof_surfaces[surface_name] = surface
                # print("debug:","analysis_all_roof_wind","wind_relation",surface.properties["wind_relation"])
        if not roof_surfaces:
            return

        # 1. Aşama: Bütün binalardaki exposed kenarların EN ÖN noktalarını kıyasla,
        # küresel rüzgarüstü sınır koordinatını (min_global_pos) bul.
        global_min_pos = float("inf")
        
        for surface in roof_surfaces.values():
            for edge in surface.edges.values():
                if not edge.get("exposed", False):
                    continue
                
                p1_pos = np.dot(np.array(edge["p1"]), w_norm)
                p2_pos = np.dot(np.array(edge["p2"]), w_norm)
                
                # Senin metodun: Sıralı liste [ön_nokta, arka_nokta]
                front_positions = sorted([p1_pos, p2_pos])  # min değer rüzgarda en öndedir
                
                if front_positions[0] < global_min_pos:
                    global_min_pos = front_positions[0]

        tol = 1e-4

        # 2. Aşama: Hem front_positions[0] hem de front_positions[1] aynı anda
        # en ön hizada olan kenarı barındıran yüzeyi global_leading yap.
        for surface_name, surface in roof_surfaces.items():
            is_leading = False
            
            for edge in surface.edges.values():
                if not edge.get("exposed", False):
                    continue

                p1_pos = np.dot(np.array(edge["p1"]), w_norm)
                p2_pos = np.dot(np.array(edge["p2"]), w_norm)
                
                # front_positions[0]: en ön nokta, front_positions[1]: en arka nokta
                front_positions = sorted([p1_pos, p2_pos])

                # ŞART: Kenarın SADECE ilk ucu değil, İKİNCİ UCU da (front_positions[1]) 
                # en ön hatta mı? (Yani kenar boydan boya ön cephede mi?)
                is_full_edge_at_front = (
                    abs(front_positions[0] - global_min_pos) <= tol and
                    abs(front_positions[1] - global_min_pos) <= tol
                )

                if is_full_edge_at_front:
                    # Nokta temaslarını elemek için kenar boyu kontrolü
                    edge_len = np.linalg.norm(np.array(edge["p2"]) - np.array(edge["p1"]))
                    if edge_len > 1e-2:
                        is_leading = True
                        break

            if is_leading:
                surface.global_leading = True
                surface.properties["global_leading"] = True
```

```python
import numpy as np

class RoofSurfaceGeometry:
    def __init__(self, name: str, coords: list):
        self.name = name
        self.coords = np.array(coords, dtype=float)

    def analyze_gable_and_hip(self, tol: float = 1e-4) -> dict:
        """
        Yüzey kenarlarını analiz ederek X veya Y düzlemine paralel 
        dik kalkan duvar kesitlerini (gable cut) ve saf kırma (pure hip) 
        durumunu tespit eder.
        """
        num_pts = len(self.coords)
        has_gable_cut = False
        gable_edges = []

        for i in range(num_pts):
            p1 = self.coords[i]
            p2 = self.coords[(i + 1) % num_pts]
            edge_vec = p2 - p1

            dx, dy, dz = edge_vec
            
            # Kenarın Z kotunda yükselip/alçalmadığını kontrol et
            has_z_slope = abs(dz) > tol
            
            # Kenar X sabit tutulup Y-Z düzleminde mi ilerliyor? (X-düzlem kalkan kenarı)
            is_const_x = abs(dx) < tol
            
            # Kenar Y sabit tutulup X-Z düzleminde mi ilerliyor? (Y-düzlem kalkan kenarı)
            is_const_y = abs(dy) < tol

            # Z kotu değişen ve X ya da Y doğrultusunda sapma yapmayan kenar kalkan kesitidir
            if has_z_slope and (is_const_x or is_const_y):
                has_gable_cut = True
                gable_plane = "X-Plane (Gable End)" if is_const_x else "Y-Plane (Gable End)"
                gable_edges.append({
                    "edge": (p1.tolist(), p2.tolist()),
                    "plane": gable_plane
                })

        # Yüzeyin saf kırma (hip) olup olmadığını belirle:
        # Eğer yüzey üçgen/yamuk ise VE dikey bir kalkan kesit kenarı içermiyorsa pure_hip'tir.
        is_trapezoidal_or_tri = num_pts in (3, 4)
        is_pure_hip = is_trapezoidal_or_tri and not has_gable_cut

        return {
            "has_gable_cut": has_gable_cut,
            "is_pure_hip": is_pure_hip,
            "gable_edges": gable_edges
        }

# --- Test ---
for plane_name in ["C1","C2","C3"]:
    coords_ = [points[k] for k in polygons[plane_name]]
    print(plane_name, coords_)
    c1 = RoofSurfaceGeometry(plane_name, coords_)
    print("analyze_gable_and_hip:", c1.analyze_gable_and_hip())
```
